# FEBRL Benchmark

Run `fuzzy_llm_matcher` against the FEBRL4 synthetic duplicate dataset (via `recordlinkage`) and report precision/recall/F1.

Requires: `pip install recordlinkage`

```octave
import sys
sys.path.insert(0, '..')
import pandas as pd
import recordlinkage
from fuzzy_llm_matcher import match_tables, evaluate_matches
```

```octave
df_a, df_b, true_links = recordlinkage.datasets.load_febrl4(return_links=True)
df_a = df_a.reset_index().rename(columns={'rec_id': 'id'})
df_b = df_b.reset_index().rename(columns={'rec_id': 'id'})
df_a['full_name'] = df_a['given_name'].fillna('') + ' ' + df_a['surname'].fillna('')
df_b['full_name'] = df_b['given_name'].fillna('') + ' ' + df_b['surname'].fillna('')
df_a.head()
```

```octave
true_pairs = pd.DataFrame(list(true_links), columns=['left_id', 'right_id'])
result = match_tables(df_a, df_b, left_on='full_name', right_on='full_name', left_id='id', right_id='id', use_llm=False)
result.head()
```

```octave
ev = evaluate_matches(result, true_pairs)
ev.to_dict()
```
