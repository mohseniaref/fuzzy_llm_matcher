# Company Name Matching Demo

Match noisy company/organization names using the bundled sample data, and inspect the reliability labels and LLM review for medium-confidence pairs.

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
from fuzzy_llm_matcher import match_tables, evaluate_matches

left = pd.read_csv('../data/sample_dirty_left.csv')
right = pd.read_csv('../data/sample_dirty_right.csv')
true_matches = pd.read_csv('../data/sample_ground_truth.csv')
left
```

```python
result = match_tables(left, right, left_on='name', right_on='name', left_id='id', right_id='id', use_llm=True)
result
```

```python
ev = evaluate_matches(result, true_matches)
ev.to_dict()
```
