# LLM Review Demo

Show how only `medium_review` pairs get sent to the LLM, using the offline `MockLLMClient`, and compare final decisions with vs. without LLM review.

```python
import sys
sys.path.insert(0, '..')
import pandas as pd
from fuzzy_llm_matcher import match_tables
from fuzzy_llm_matcher.llm_review import MockLLMClient, build_prompt
```

```python
left = pd.DataFrame({'id': [1,2,3], 'name': ['Acme Corp', 'Statista Strategy', 'SkyGeo NL']})
right = pd.DataFrame({'id': [10,11,12], 'name': ['Acme Widgets Inc', 'Statista Strategy GmbH & Co. KG', 'Sky Geo B.V.']})

without_llm = match_tables(left, right, left_on='name', right_on='name', left_id='id', right_id='id', use_llm=False)
without_llm
```

```python
with_llm = match_tables(left, right, left_on='name', right_on='name', left_id='id', right_id='id', use_llm=True, llm_client=MockLLMClient())
with_llm
```
